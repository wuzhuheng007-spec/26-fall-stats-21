## Persistence

Variables usually live only for the current Python session. Restarting the kernel discards them, even if the notebook still displays saved output.

To keep data between sessions, we need **persistent storage**. This lecture introduces reading and writing files to save data.

## Opening a file

`open('output.txt', 'w', encoding='utf-8')` opens a text file for writing.

**Opening an existing file in `'w'` mode immediately erases its contents**, even before you call `write()`. If the file does not exist, Python creates it, provided its parent directory exists and you have permission.

| Mode | Purpose |
|---|---|
| `'r'` | Read an existing file (the default). |
| `'w'` | Write, replacing existing contents. |
| `'a'` | Append to the end, preserving existing contents. |
| `'x'` | Create a new file; raise an error if it already exists. |

We specify UTF-8 so text is read and written consistently across computers.

```python
fout = open('output.txt', 'w', encoding='utf-8')
```

The function `open()` returns a file object that allows you to write to the file.

```python
line1 = "This is the first line of text.\n"
fout.write(line1)
```

For a text file, `write()` returns the number of characters written. It does not add a newline; the `\n` in `line1` supplies one.

A second `write()` on this same open file continues at its current position, after the first line. Reopening the file with `'w'` would erase what we already wrote; use `'a'` to reopen it for appending.

```python
line2 = "This will be the second line of text.\n"
fout.write(line2)
```

When you are done writing, you should close the file.

```python
fout.close()
```

Closing the file flushes buffered output and releases the file resource. Do not rely on garbage collection to close files promptly. For normal file work, prefer the `with` form below.

## Closing files automatically with `with`

A `with` statement opens the file for the indented block and closes it when the block exits, including when an exception occurs. This is the usual way to work with files.

```python
# opening a file and writing lines to it:
with open('output.txt', 'w', encoding='utf-8') as file:
    file.write(line1)
    file.write(line2)
```

```python
# Each line already includes a newline, so avoid adding a second one.
with open('output.txt', 'r', encoding='utf-8') as file:
    for line in file:
        print(line, end='')
```

## Formatting strings

For new code, f-strings are a convenient way to insert expressions directly into text: `f'{x}'`. The `str.format()` method remains useful, especially for a format string stored separately from its values.

Both support a **format specification** after `:` inside a replacement field. Formatting creates text for display; it does not change the original number.

Reference: [Python format specifications](https://docs.python.org/3/library/string.html#format-specification-mini-language).

```python
x = 12.34
y = '{} is my favorite number'.format(x)
print(y)
```

```python
# equivalent f-string
# when using the f-string, place an expression (often a variable name) inside the braces
yfstring = f'{x} is my favorite number'
print(yfstring)
```

```python
x = 12.34
x2 = 37.8
# you can use multiple replacement fields and Python inserts values in argument order
y = '{} is my favorite number and {} is my second favorite number'.format(x, x2)
print(y)
```

```python
x = 12.34
x2 = 37.8
# you can set index values for replacement fields and python will insert values
# using the zero-based positions of the arguments passed to .format()
y = '''{1} is my favorite number and {0} is my second favorite number.
Again, my favorite number is {1}.'''.format(x, x2)
print(y)
```

```python
x = 321
x2 = 567
yfstring = f'''{x} is my favorite number and {x2} is my second favorite number. 
Again, my favorite number is {x}.'''
print(yfstring)
```

```python
# :g uses general format: fixed-point or scientific notation as appropriate.
# The default precision is 6 significant digits; .8g requests 8.
# Trailing zeros are normally removed, so fewer digits may be displayed.
x = 0.000123456789
x2 = 0.0000123456789
y = '{0:g} is my favorite number. I also like {1:.8g}.'.format(x, x2)
print(y)
```

```python
# f-strings use the same Python format specifications
yfstring  = f'{x:g} is my favorite number. I also like {x2:.8g}.'
print(yfstring)
```

```python
# :f uses fixed-point format, with 6 digits after the decimal by default.
# .8f displays 8 digits after the decimal, rounding as needed.
x = 0.000123456789
x2 = 0.0000123456789
y = '{0:f} is my favorite number. I also like {1:.8f}'.format(x, x2)
print(y)
```

```python
# alternative method using + operator
x = 12.345
age = 85
y = 'My age is ' + str(age) + ', and ' + str(x) + ' is my favorite number'
print(y)
```

```python
# Placeholder indices are optional when values are used in argument order.
y = 'My name is {:s}, I am {:g} years old, and {:g} is my favorite number'.format("Joe Bruin", 100, 83)
print(y)
```

### Aligning text

Use `<` for left alignment, `>` for right alignment, and `^` for centering. The width is a **minimum total field width**, including the text, rather than the number of extra spaces. Longer values are not truncated.

```python
'{:<30}'.format('left aligned')
```

```python
'{:>30}'.format('right aligned')
```

```python
'{:^30}'.format('centered')
```

```python
# f string Left/center/right alignment in 12 columns
s = "UCLA"
print(f"|{s:<12}|{s:^12}|{s:>12}|")
print(f"|{s:>12}|{s:^12}|{s:<12}|")
```

### Common f-string formats

```python
# Display with a thousands separator and 2 decimal places
amount = 1234.5
print(f"${amount:,.2f}")
```

```python
# A percentage format multiplies the displayed value by 100 and adds %.
ratio = 0.0739
print(f"{ratio:.1%}")
```

```python
# Zero-padded integers
i = 42
print(f"{i:05d}")
```

```python
# Display hours:minutes:seconds with a minimum width of 2 digits per value
h = 1
m = 2
s = 7
d = "PM"
print(f'{h:02d}:{m:02d}:{s:02d} {d}')
```

```python
# Columnar table built from tuples
rows = [("penguin", 3), ("albatross", 12)]
for animal, n in rows:
    print(f"{animal:<12} {n:>5d}")
```

## Catching exceptions

Python **raises an exception** when an operation fails. A `try` block lets you attempt an operation; a matching `except` block handles a particular kind of failure.

The next examples assume that no file named `bad_file` exists. The first cell intentionally raises `FileNotFoundError`.

```python
# Intentional error: this file should not exist in the practice folder.
with open('bad_file', encoding='utf-8') as fin:
    print(fin.read())
```

```python
try:
    with open('bad_file', encoding='utf-8') as fin:
        print(fin.read())
except FileNotFoundError:
    print('The file was not found, but the exception was caught.')
```

Python runs the `try` block. If no exception occurs, it skips `except`. If an exception occurs, it skips the rest of the `try` block and looks for a handler with a **matching exception type** (or one of its parent types).

An unmatched exception propagates to the caller. The next cell intentionally demonstrates that `except KeyError` does not catch `FileNotFoundError`.

```python
# Intentional error: the handler does not match the exception raised.
try:
    with open('bad_file', encoding='utf-8') as fin:
        print(fin.read())
except KeyError:
    print('This catches KeyError, not FileNotFoundError.')
```

Catch the specific exceptions you can handle. See the [Python exception hierarchy](https://docs.python.org/3/library/exceptions.html#exception-hierarchy).

## Saving data with JSON

**JSON** (JavaScript Object Notation) is a common text format for exchanging and saving simple data. Python's `json` module supports dictionaries, lists, strings, numbers, booleans, and `None`.

It does not preserve arbitrary Python objects or every Python type: tuples load as lists, dictionary keys become strings, and sets and NumPy arrays need conversion. Use string keys when you want a dictionary to round-trip unchanged.

`json.dump()` writes JSON to a text file; `json.load()` reads it back. Standard JSON has no `NaN` or infinity values; `allow_nan=False` rejects these when writing.

```python
import json

# Let's say we have a dictionary we want to save
my_data = {
    "name": "Joe Bruin",
    "age": 100
}
```

```python
my_data
```

```python
# We use json.dump() to write the object to a file
with open('data.json', 'w', encoding='utf-8') as file:
    json.dump(my_data, file, allow_nan=False)
```

Now that the data is saved in `data.json`, we can retrieve it later using `json.load()`.

```python
# We use json.load() to read the file back into a Python object
with open('data.json', 'r', encoding='utf-8') as file:
    loaded_data = json.load(file)

print(type(loaded_data))
print(loaded_data)
```

```python
my_data
```

```python
# A JSON document contains one top-level value. To save several objects,
# collect them in a dictionary (as here) or a list. Repeated json.dump() calls
# to the same file would not produce one valid JSON document.
my_list = [1, 2, 3, 4, 5]
my_string = "This is a string."
my_number = 42

# Package everything into a single dictionary
data_to_save = {
    "my_data": my_data,
    "my_list": my_list,
    "my_string": my_string,
    "my_number": my_number
}

# Save that one dictionary to the file
with open('data.json', 'w', encoding='utf-8') as file:
    # Adding indent=4 makes the file easily readable for humans!
    json.dump(data_to_save, file, indent=4, allow_nan=False)
```

```python
# when you are ready to read the file back, use json.load()
# to read the entire dictionary back into memory, and then pull out the variables you want using their keys
with open('data.json', 'r', encoding='utf-8') as file:
    # This loads the entire dictionary back into memory
    loaded_data = json.load(file)

print(loaded_data)

# Now you can just pull your variables back out using their keys
loaded_dict = loaded_data["my_data"]
loaded_list = loaded_data["my_list"]
loaded_string = loaded_data["my_string"]
loaded_number = loaded_data["my_number"]
```

## JSON Lines: one record per line

**JSON Lines** (`.jsonl`) stores one complete JSON value per line. A value can be a dictionary, list, string, number, boolean, or `null`; it need not be an object/dictionary. Blank lines are not valid records.

Use `json.dumps()` to produce a string, add `\n`, and use `json.loads()` to parse a line. Use UTF-8. Unlike one JSON document, a JSON Lines file can be processed record by record without loading every record at once.

```python
list2 = [[1,2,3], "This is a string."]
```

```python
with open('data.jsonl', 'w', encoding='utf-8') as file:
    # Write each object as a JSON string on its own line
    file.write(json.dumps(my_list, allow_nan=False) + '\n')
    file.write(json.dumps(list2, allow_nan=False) + '\n')
    file.write(json.dumps(my_string, allow_nan=False) + '\n')
    file.write(json.dumps(my_number, allow_nan=False) + '\n')
```

```python
with open('data.jsonl', 'r', encoding='utf-8') as file:
    # Read the file line by line and parse each one
    loaded_item1 = json.loads(file.readline())
    loaded_item2 = json.loads(file.readline())
    loaded_item3 = json.loads(file.readline())
    loaded_item4 = json.loads(file.readline())
    # All four records have now been read. At EOF, readline() returns an empty string,
    # which is not valid JSON. The loop below handles an unknown number of records.
```

```python
loaded_item1
```

```python
# Collecting records in a list is convenient, but uses memory for the whole dataset.
loaded_items = []
with open('data.jsonl', 'r', encoding='utf-8') as file:
    for line in file:
        loaded_item = json.loads(line)
        loaded_items.append(loaded_item)
for item in loaded_items:
    print(item)
```

```python
loaded_items
```

```python
# For a large file, process one record at a time instead of storing all records.
with open('data.jsonl', 'r', encoding='utf-8') as file:
    for line in file:
        item = json.loads(line)
        print(item)  # Replace with the work you need to do for each record.
```
