## NumPy arrays

- An `ndarray` is a rectangular, multidimensional container whose values can usually be modified in place.
- Each array has one `dtype`, which determines how its elements are stored.
- NumPy converts mixed inputs to a common dtype when possible. An object array can hold references to different Python objects, but our examples use ordinary numeric arrays.

### Arrays in higher dimensions

A rectangular list of equal-length lists produces a 2D array, which we can use to represent a matrix. Lists nested to greater depth can create more dimensions.

```python
d = np.array( [ [1,2,3] , [4,5,6] ] ) 
print(d)
```

`d.ndim` gives the number of dimensions, and `d.shape` gives the length of each axis. Here, `d.shape` is `(2, 3)`: 2 rows and 3 columns.

A regular 2D array needs the same number of columns in every row. The next cell intentionally fails because its inner lists have different lengths. For ragged data, keep a list of lists or explicitly use an object array; it will not behave like a rectangular numeric matrix.

```python
# Intentional error: the inner lists have different lengths.
e = np.array([ [1,2,3],[4,5] ])
print(e)
```

## Making arrays of random numbers

Create a random-number generator with `np.random.default_rng()`, the recommended API for new NumPy code. It currently uses **PCG64** by default. The older `np.random` global API uses a different generator, MT19937 (Mersenne Twister).

A fixed seed makes examples repeatable when you use the same environment and sequence of calls. Each call advances the generator's state, so rerunning only a sampling cell produces new values. Recreate the generator to restart its sequence; exact streams are not guaranteed across NumPy versions.

```python
# create a new random number generator with a fixed seed for reproducibility
rng = np.random.default_rng(seed=1)
```

```python
print(rng.random(5))  # 5 draws from the uniform distribution on [0, 1)
```

```python
rng.standard_normal(5)
# 5 draws from a standard normal distribution (mean 0, standard deviation 1)
```

```python
rng.standard_normal((5, 2)) # you can provide a tuple of dimensions
```

```python
rng.normal(loc=10, scale=3, size=(2, 4))
# A 2 x 4 array from a normal distribution with mean 10 and standard deviation 3.
# scale is the standard deviation, not the variance.
# A small random sample will not have exactly the specified mean and SD.
```

```python
rng.integers(0, 10, 20) 
# select random integers from 0 inclusive to 10 exclusive
# and return 20 values
```

```python
# Simulate 50 rolls of a fair six-sided die: 1 through 6, with repeats allowed.
rng.integers(1, 7, size=50)
```

More distributions and methods: [NumPy random sampling](https://numpy.org/doc/stable/reference/random/index.html).

## Array sequences: `arange`

`np.arange(start, stop, step)` makes regularly spaced values starting at `start` and stopping before `stop`. With one argument, `start=0` and `step=1`.

Use it especially for integer steps. Floating-point steps can cause rounding surprises, including an unexpected final value; prefer `linspace` when you want a specific number of evenly spaced values.

```python
range(0, 10, 2) # range object in regular python
```

```python
list(range(0, 10, 2))
```

```python
np.arange(0, 100, 5)
```

## Array sequences: `linspace`

`np.linspace(start, stop, num)` produces `num` evenly spaced values, including both endpoints by default (when `num >= 2`). Here `num` is a **count**, not a step size.

With `endpoint=False`, `stop` is excluded and NumPy adjusts the spacing to still produce `num` values.

```python
np.linspace(0, 100, 11)
```

```python
np.linspace(0, 100, 10)
```

```python
np.linspace(0, 100, 10, endpoint = False)  
# optional parameter endpoint to exclude the stop value
```

```python
np.linspace(0, 100, 9, endpoint = False) 
# if you use the endpoint argument, the last number in the array will depend on the output length
```

### A common dtype

When creating an array from mixed values, NumPy infers a dtype that can represent the inputs according to its conversion rules. For example, integers and floats often become floats; adding strings can produce a string array.

This conversion is not a promise to preserve all precision or meaning. Check `.dtype`, and specify `dtype=` when the representation matters.

```python
b = np.array([1, 2, 3.0, False, True])
print(b) # the 3.0 is a float and will upcast (coerce) other values to floats
```

```python
c = np.array([1, 2, "3", True, False]) # upcast (coerced) to strings
print(c)
```

### Assignment does not change an array's dtype

Assigning a float to an integer array can discard its fractional part. Convert the array to a floating dtype first when you need decimal values.

```python
integer_values = np.array([1, 2, 3])
integer_values[0] = 1.9
print(integer_values)  # [1 2 3]: assignment converted 1.9 to an integer

float_values = integer_values.astype(float)
float_values[0] = 1.9
print(float_values)
```

## Reshaping and transposing arrays

- `np.reshape(array, new_shape)` and `array.reshape(new_shape)` return an array with the requested shape. The total number of elements must stay the same.
- Reshaping preserves element order (row-major by default) and returns a view when possible, otherwise a copy. It does not change the original array's shape.
- `array.T` is a transpose **attribute**. For 2D arrays, it swaps rows and columns. The returned array shares data with the original.

```python
j = np.arange(0,12,1)
print(j) # j is one dimensional
```

```python
k = np.reshape(j, (3,4))  # default order fills rows; the result shares data with j here
print(k)
```

```python
j # j still has shape (12,); j and k share the underlying values
```

```python
j.reshape(4,3)  # you can also call the method reshape() on the array j
```

```python
j # j is left unchanged here as well
```

```python
print(k)
```

```python
print(k.T)  # the transpose of k
```

```python
print(k) # accessing k.T does not change k; modifying the transpose can change shared values
```

```python
# can combine the above methods and steps into one:
l = np.arange(0,12,1).reshape((3,4)).T
# create a-range >> reshape >> transpose
print(l)
```

```python
j
```

Use `-1` for **one** dimension to have NumPy infer its length. The known dimensions must divide the total size evenly.

```python
j.reshape((3, -1))  # using -1 for a dimension will ask python to figure out the number to use for that dimension
```

```python
j.reshape((-1, 4))
```

```python
j.reshape((2, -1, 2)) # two sheets, unknown number of rows, 2 columns
```

```python
y = np.arange(0,12, 1)
print(y)
```

```python
y.shape
```

```python
print(y.T) # the transpose of a one dimensional array doesn't suddenly give it a second dimension
```

```python
z = np.reshape(y, (1,12)) # the array now has two dimensions
print(z)
```

```python
z.shape
```

```python
print(z.T)  # with two dimensions, the transpose become a column
```

```python
z.T.shape
```

### Flattening an array

- `ravel()` returns a 1D array, sharing memory when possible and copying otherwise.
- `flatten()` returns a 1D copy.

For this contiguous array, `C.ravel()` shares memory with `C`. The `C += 1` operation modifies the existing data, so the view changes too.

```python
C = np.arange(6).reshape(2,3)
C
```

```python
C_unraveled = C.ravel()      # a view in this example
C_unraveled
```

```python
C_flattened = C.flatten()    # copy
C_flattened
```

```python
C += 1
C
```

```python
C_unraveled
```

```python
C_flattened
```

```python
print(np.shares_memory(C, C_unraveled))  # True
print(np.shares_memory(C, C_flattened))  # False
```

# Indexing and slicing arrays

Indices are zero-based; negative indices count from the end. As with lists, slices include the start and exclude the stop. Unlike list slices, basic NumPy slices share data.

```python
y = np.arange(0,12, 1)
print(y)
```

```python
y[4]
```

```python
y.shape
```

```python
y[4:6]
```

You can add a second colon: `array[start:stop:step]`. The stop index is excluded. The step can also be negative to traverse in reverse.

```python
y[1:8:3]
```

```python
np.arange(100)[:100:2] # to get even values
```

```python
np.arange(0,100,2)
```

Subsetting and slicing higher dimensional arrays is similar, and uses a comma to separate subsetting instructions for each dimension.

```python
z = np.reshape(y, [3,4])
print(z)
```

```python
print(z[1,2]) # returns what is at row index 1, col index 2
```

```python
type(z[1,2]) # Integer indexing on every axis returns a NumPy scalar, not an ndarray.
```

```python
print(z)
z[0:2, 0:2] # note the type remains a numpy array
```

```python
print(z[2, :]) # returns row at index 2
```

```python
print(z[2, ])
```

```python
z[2, :].shape  # the shape is one dimensional
```

```python
print(z[:,2]) # returns column at index 2
```

```python
z[:,2].shape # shape is one dimensional
```

Integer indexing removes an axis; a slice preserves it. Compare these 2D selections with the 1D results above.

```python
print(z[2:3, :].shape)  # one row, still 2D: (1, 4)
print(z[:, 2:3].shape)  # one column, still 2D: (3, 1)
```

## Basic slices are views

A slice such as `z[:2, :2]` shares the original array's data. In-place changes are visible through both arrays. Rebinding a variable to a different array does not redirect existing views.

This differs from selecting with a list of indices or a Boolean mask, which returns a copy (covered next lecture).

```python
z = np.arange(12).reshape([3,4])
print(z)
```

```python
# we use numpy array slicing to create z_sub, the top left corner of z
z_sub = z[:2, :2]
print(z_sub)
```

```python
# I modify the first element of z to be 99.
z[0,0] = 99
```

```python
print(z_sub)  # z_sub is updated, even though we never redefined it
```

```python
z
```

```python
z = np.arange(15).reshape([3,5]) # here z gets redefined to an entirely new object
# we are not modifying the object that used to be called z
# we created a new object, and the name z now points to the new object
```

```python
z
```

```python
print(z_sub)  # the view z_sub still points to the object formerly known as z, which was not modified 
```

## Use `.copy()` for independent array data

For these numeric arrays, modifying the original will not affect a copy, and modifying a copy will not affect the original.

```python
print(z)
```

```python
z_sub_copy = z[:2, :2].copy()
print(z_sub_copy)
```

```python
z[0,0] = 55 # modify the first element of z
```

```python
print(z_sub_copy) # the copy remains unaffected by the change
```

```python
print(z)
```

## A view can modify the shared data

Changes work in both directions: assigning to the view also changes the corresponding values in the original array.

```python
z = np.arange(12).reshape((3,4))
print(z)
```

```python
view = z[:2,:2]
```

```python
view[0,0] = 99
```

```python
view
```

```python
z
```

```python
type(view) # A view is itself an ndarray, with the usual array methods and attributes.
```

```python
view.T
```

```python
view.T.reshape((4,))  # reshaping this noncontiguous view requires a copy
```

```python
view # Merely accessing .T does not modify values; assigning through it can.
```

```python
view[0,0] = 1000
```
