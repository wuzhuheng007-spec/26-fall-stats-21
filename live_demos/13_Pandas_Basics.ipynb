## From arrays to tables

A NumPy array has one `dtype`. A pandas **DataFrame** is a two-dimensional table
with row and column labels. Its columns can have different dtypes.

Selecting one column gives a **Series**: a one-dimensional object with labels
and a dtype. A Series can use NumPy storage or other array types; a DataFrame is
not simply a single NumPy matrix.

Use the conventional import:

```python
import pandas as pd

pd.__version__
```

## 4. Labels and positions: a small Avengers table

Here is one useful way to construct a DataFrame: give a dictionary whose keys
are column names and whose values are equal-length lists.

We will use character names as row labels to make the distinction between labels
and positions visible. The ages are illustrative values for this fictional example.

```python
avengers = pd.DataFrame(
    {
        "actor": ["Robert Downey Jr.", "Chris Evans", "Scarlett Johansson", "Chris Hemsworth"],
        "hero name": ["Iron Man", "Captain America", "Black Widow", "Thor"],
        "age": [42, 104, 28, 1493],
    },
    index=["Tony Stark", "Steve Rogers", "Natasha Romanoff", "Thor"],
)
avengers
```

### `.loc[rows, columns]` selects by label

Supply a row label and a column label to get a value. A colon `:` means all labels
along that dimension. Lists of both row and column labels keep the result as a DataFrame.

```python
avengers.loc["Thor", "age"]
```

```python
avengers.loc[["Thor", "Natasha Romanoff"], ["hero name", "age"]]
```

```python
avengers.loc[:, "age"]  # all rows, one column: a Series
```

### `.iloc[rows, columns]` selects by position

Positions start at zero. Positional slices exclude the stop, just as in Python
lists and NumPy arrays.

Use `.loc` when you mean labels; use `.iloc` when you mean positions. For example,
`.loc[1]` would mean the row labeled 1, even if it is not the second row.

```python
avengers.iloc[0]  # first row, returned as a Series
```

```python
avengers.iloc[:2, :2]  # first two rows and first two columns
```

## 5. Filter rows with a Boolean condition

Back to Titanic: which records are for first-class passengers?

A comparison produces one Boolean value per row. The condition below is a
Series whose index matches the table it came from.

```python
first_class = titanic["pclass"] == 1
first_class.head(10)
```

```python
titanic.loc[first_class, ["name", "pclass", "fare"]].head()
```

### Put the condition directly inside `.loc`

The general pattern is `df.loc[row_condition, column_names]`.
Build the condition from the same DataFrame you are filtering.

```python
titanic.loc[titanic["pclass"] == 1, ["name", "pclass", "fare"]].head()
```

### Combine conditions

As with NumPy, use `&` for elementwise AND, `|` for elementwise OR, and `~` for NOT.
Put parentheses around each comparison. Python's `and` and `or` do not combine
these Series element by element.

```python
# Female passengers in first class.
titanic.loc[
    (titanic["sex"] == "female") & (titanic["pclass"] == 1),
    ["name", "sex", "pclass", "fare"],
].head()
```

```python
# Passengers in either first or second class.
titanic.loc[
    (titanic["pclass"] == 1) | (titanic["pclass"] == 2),
    ["name", "pclass", "fare"],
].head()
```

### Membership: `.isin()`

`isin()` tests membership in a collection of values. It is often clearer than
writing a long sequence of OR conditions.

```python
titanic.loc[
    titanic["pclass"].isin([1, 2]),
    ["name", "pclass", "fare"],
].head()
```

### An age condition uses the recorded ages

The following selects passengers with recorded ages of at least 18.
In this CSV, missing ages are floating-point `NaN`; their comparison to 18 is
`False`, so those rows are excluded.

An excluded passenger is therefore not necessarily a child. We will return
to missingness and its effect on an analysis in lecture 14.

```python
titanic.loc[titanic["age"] >= 18, ["name", "age"]].head()
```

## 6. Create variables with column operations

Arithmetic on columns works element by element. We can save the result as a new
column using `df["new_name"] = expression`.

`sibsp + parch` counts recorded siblings/spouses and parents/children aboard.
Call this **relatives_aboard**; it does not include the passenger.

```python
titanic["relatives_aboard"] = titanic["sibsp"] + titanic["parch"]
titanic[["name", "sibsp", "parch", "relatives_aboard"]].head()
```

```python
titanic["no_recorded_relatives"] = titanic["relatives_aboard"] == 0
titanic[["name", "relatives_aboard", "no_recorded_relatives"]].head()
```

### What did we just change?

We added columns while keeping the same passenger rows. These expressions use
vectorized column operations; no loop over passengers is needed.

**Interpretation:** `no_recorded_relatives` means zero recorded relatives in the
`sibsp` and `parch` categories. It does not establish that a passenger had no
friends or other companions aboard.

Both source columns are complete in this file. If they contained missing values,
we would need to consider what an unknown count means before creating the flag.

```python
titanic.shape
```

### Modify the intended table directly

Use a single `.loc[rows, columns] = value` operation for an update to selected
cells. We will change an illustrative value in the Avengers table.

Avoid chained assignment such as `df["column"][condition] = value`. In pandas 3,
that does not update the original DataFrame. A selected subset behaves independently
when modified; the optional examples show this rule in action.

```python
avengers.loc["Thor", "age"] = 1500
avengers.loc[["Thor"], ["hero name", "age"]]
```

## 7. Sort whole rows

`sort_values()` orders the table using one or more columns. The values within
each passenger record stay together.

Ascending order is the default. Use `ascending=False` for descending order.
Sorting returns a result; assign it to a variable if you want to keep it.

```python
titanic.sort_values("fare").loc[:, ["name", "pclass", "fare"]].head()
```

```python
highest_fares = titanic.sort_values(
    ["fare", "passengerid"], ascending=[False, True]
)
highest_fares[["passengerid", "name", "pclass", "fare"]].head()
```

### Sorting by more than one column

Above, fare is the main sort key. Passenger ID breaks fare ties in increasing
order, making the displayed order predictable.

The original `titanic` still has its original row order. `highest_fares` has the
same rows in a new order, with their existing index labels.

```python
titanic[["passengerid", "fare"]].head()
```

## 8. Put the operations together

**Question:** Among third-class passengers who survived and had no recorded
relatives aboard, which ten records have the highest fares?

1. Filter for all three conditions.
2. Keep `passengerid`, `name`, `sex`, `age`, `fare`, and `relatives_aboard`.
3. Sort by descending fare, then ascending passenger ID to break ties.
4. Display the first ten rows.

Before coding, describe the steps using the familiar ideas **filter**, **select**,

### One solution: name the intermediate result

The Boolean column can be used directly as a condition; it already contains
`True` or `False` values.

```python
selected_passengers = titanic.loc[
    (titanic["pclass"] == 3)
    & (titanic["survived"] == 1)
    & titanic["no_recorded_relatives"],
    ["passengerid", "name", "sex", "age", "fare", "relatives_aboard"],
]

fare_order = selected_passengers.sort_values(
    ["fare", "passengerid"], ascending=[False, True]
)
fare_order.head(10)
```

### Read the result carefully

- How many records met the conditions, before taking the first ten?
- Why are some displayed ages missing, even though our code ran successfully?
- Did selecting columns change the number of rows? Did sorting?
- Could this filtered list establish whether having relatives affected survival?

We did not filter on age, so unknown ages can remain. This is a selected list of
survivors, not a comparison of survival rates. Lecture 14 will build comparisons
with explicit group sizes and denominators.

## The verbs you already know

| Task | R/dplyr idea | pandas pattern used today |
|---|---|---|
| Select columns | `select()` | `df[["col1", "col2"]]` |
| Filter rows | `filter()` | `df.loc[condition, columns]` |
| Create a variable | `mutate()` | `df["new"] = expression` |
| Order rows | `arrange()` | `df.sort_values(...)` |

These are conceptual translations, not identical function names. In particular,
pandas' method named `.filter()` selects labels; it is not dplyr's row-condition filter.

Next: missingness, duplicate records, and grouped summaries. Later we will use
these same operations in Polars and prepare tables for visualization.

### chain operations

Methods such as `sort_values()` and `head()` return objects that support the next
operation. Parentheses let us write the steps on separate lines. This reproduces
the cumulative exercise without naming its intermediate tables.

```python
(
    titanic.loc[
        (titanic["pclass"] == 3)
        & (titanic["survived"] == 1)
        & titanic["no_recorded_relatives"],
        ["passengerid", "name", "sex", "age", "fare", "relatives_aboard"],
    ]
    .sort_values(["fare", "passengerid"], ascending=[False, True])
    .head(10)
)
```

### filter text values

The `.str` accessor applies string operations to a Series. Here, `regex=False`
asks for a literal substring, and `na=False` treats missing names as nonmatches.

```python
titanic.loc[
    titanic["name"].str.contains("Williams", regex=False, na=False),
    ["name", "age", "pclass"],
]
```

### label slices versus position slices

With these ordered, unique row labels, a `.loc` label slice includes both
endpoints. An `.iloc` position slice excludes the stop.

```python
avengers.loc["Tony Stark":"Natasha Romanoff", ["hero name", "age"]]
```

```python
avengers.iloc[0:2, :]  # only the first two rows
```

### Series align by label

Constructing a Series from a dictionary uses its keys as labels. When a Series
is assigned to a DataFrame column, pandas matches row labels, rather than merely
using the Series' current order.

This is one reason to keep track of labels when creating columns from another table.

```python
# The dictionary order deliberately differs from the Avengers row order.
appearances = pd.Series({
    "Thor": "Thor",
    "Natasha Romanoff": "Iron Man 2",
    "Steve Rogers": "Captain America: The First Avenger",
    "Tony Stark": "Iron Man",
})

avengers_with_films = avengers.copy()
avengers_with_films["first_MCU_film"] = appearances
avengers_with_films[["hero name", "first_MCU_film"]]
```

### selecting a subset in pandas 3

A selected Series behaves independently when you modify it. To change the
DataFrame, assign directly to that DataFrame instead.

This differs from the shared-data behavior of the NumPy basic slices studied in
lecture 11. Simply writing `other = avengers` would still give two names for the
same object; selecting a column is a different operation.

```python
age_subset = avengers["age"]
age_subset.loc["Thor"] = 1600

print("Selected Series:", age_subset.loc["Thor"])
print("Original DataFrame:", avengers.loc["Thor", "age"])
```

### converting a Series to a NumPy array

`to_numpy()` returns an ndarray without the Series' index labels. Its dtype may
differ from the Series' dtype. A mixed row like the one below produces an ordinary
`object` array holding Python objects; it is not a structured array with named fields.

```python
thor_values = avengers.loc["Thor"].to_numpy()
print(thor_values)
print(type(thor_values), thor_values.dtype)
```
