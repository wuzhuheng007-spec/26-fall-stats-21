## Concatenating arrays

`np.concatenate` joins arrays along an **existing** axis. Inputs must have the same number of dimensions and matching lengths on every axis except the one being joined.

```python
x = np.arange(4)
y = np.arange(100, 104)
print(x)
print(y)
```

```python
x.shape
```

```python
np.concatenate([x,y])
```

`np.concatenate` defaults to `axis=0`. Axes are numbered from zero; a 1D array has only axis 0.

```python
np.concatenate([x,y], axis = 0)
```

The next cell intentionally tries `axis=1` on 1D arrays and raises `AxisError`. To stack them as rows, first introduce a second dimension or use `np.vstack`.

```python
np.concatenate([x, y], axis=1)  # Intentional error: these arrays have no axis 1.
```

```python
x.shape # you can't use axis with index 1, because axis index 1 does not exist
```

```python
np.vstack([x,y])   # vstack will vertically stack unidimensional arrays
```

```python
x.reshape(1,4)
```

```python
print(y.reshape(1,4))
```

```python
np.concatenate([x.reshape(1,4), y.reshape(1,4)], axis = 0)
```

For a 2D array, `axis=0` joins rows and increases the row count; the column counts must match. `axis=1` joins columns and increases the column count; the row counts must match.

```python
print(np.concatenate([x.reshape(1,4), y.reshape(1,4)], axis = 1))
```

```python
xm = np.arange(6).reshape((2,3))
ym = np.arange(100,106,1).reshape((2,3))
print(xm)
print(ym)
```

```python
xm.shape
```

```python
ym.shape
```

```python
print(np.concatenate([xm,ym])) # default behavior concatenates on axis 0
```

```python
print(np.concatenate([xm,ym], axis = 0))  
# axes are reported as rows, then columns.
# concatenating along axis 0 will concatenate along rows
```

```python
print(np.concatenate([xm,ym], axis = 1))
# concatenating along axis 1 will concatenate along columns
```

```python
np.vstack([xm, ym])
```

```python
np.hstack([xm, ym])
```

For 2D arrays, `vstack` joins along axis 0 and `hstack` joins along axis 1, with the same shape requirements as `concatenate`.

For 1D inputs, `vstack` first turns each input into a row; `hstack` joins the values into one longer 1D array. Use `np.stack` when you specifically want to create a new axis from equal-shaped arrays.

For 1D arrays of the same length, `x @ y` is their dot product, a scalar. For 2D arrays, `@` is matrix multiplication: `(m, n) @ (n, p)` produces shape `(m, p)`.

## Summaries for 2D arrays

The `axis` is the dimension that gets reduced:

| For an array of shape `(3, 4)` | Operation | Output shape |
|---|---|---|
| `axis=None` (default) | Combine all 12 entries | Scalar |
| `axis=0` | Reduce the 3 rows; one result per column | `(4,)` |
| `axis=1` | Reduce the 4 columns; one result per row | `(3,)` |

With `keepdims=True`, a reduced axis remains with length 1, which is useful for broadcasting.

```python
rng = np.random.default_rng(1)
# Alternative data: M = rng.random((3, 4))
M = np.arange(12)
rng.shuffle(M)  # shuffle the 1D values in place before reshaping
M = M.reshape((3, 4))
print(M)
```

```python
sum(M) # regular sum function
```

```python
np.sum(M) # np.sum function
```

```python
print(M)
```

```python
np.sum(M, axis=0)  # reduce rows: one total per column, shape (4,)
```

```python
np.sum(M, axis=1)  # reduce columns: one total per row, shape (3,)
```

```python
np.min(M, axis=0)  # one minimum per column
```

```python
print(M)
```

### Population versus sample standard deviation

`np.std` defaults to `ddof=0`: divide the sum of squared deviations by the number of observations, then take the square root. This describes the spread of the values treated as a complete population.

Use `ddof=1` for the usual sample standard deviation, which divides by `n - 1`. For independent, identically distributed observations with finite variance, this correction makes the **variance** estimate unbiased; the standard deviation itself is still generally biased. At least two observations are needed for `ddof=1`.

```python
np.std(M)  # ddof=0; treat all 12 entries as one population
```

```python
np.std(M, ddof=1)  # treat all 12 entries as a sample
```

```python
np.std(M, axis=0)  # population SD separately for each column (3 entries each)
```

```python
np.std(M, axis=0, ddof=1)  # sample SD separately for each column
```

```python
np.mean(M, axis=1)  # one mean per row
```

## Summaries for higher-dimensional arrays

For shape `(2, 3, 4)`, axis 0 indexes sheets, axis 1 rows, and axis 2 columns. Reducing one axis removes that dimension unless `keepdims=True`.

```python
rng = np.random.default_rng(1)
A = np.ones(24)
rng.shuffle(A)  # retained demonstration: shuffling identical values has no visible effect
A = A.reshape((2, 3, 4))  # 2 sheets, 3 rows, 4 columns
print(A)
```

```python
np.sum(A, axis=0)  # combine 2 sheets -> shape (3, 4), each total is 2
```

```python
np.sum(A, axis=1)  # combine 3 rows -> shape (2, 4), each total is 3
```

```python
np.sum(A, axis=2)  # combine 4 columns -> shape (2, 3), each total is 4
```

```python
# Extra practice: distinct entries make it easier to trace each sum.
A_labeled = np.arange(24).reshape((2, 3, 4))
print(A_labeled)
print(np.sum(A_labeled, axis=0))
print(np.sum(A_labeled, axis=1))
print(np.sum(A_labeled, axis=2))
```

## Working with NaN

`NaN` means "not a number." It can result from an undefined numerical operation and is also commonly used to represent missing numeric data.

Ordinary integer NumPy dtypes cannot store NaN. Floating-point dtypes can (as can components of complex values). `np.nan` is a floating-point value. Infinity, `np.inf`, is a separate value.

```python
x = np.nan  # direct creation of nan
print(x)
print(type(x))
```

```python
y = float("inf")  # y is the float representation of infinity
print(y / y)  # these calculations will yield a nan result
print(y - y)
```

```python
7 - y
```

```python
4 / y
```

```python
np.sum([x, 2])
```

```python
np.nansum([x, 2])  # ignore NaN entries; here the result is 2
```

```python
np.nanmean([x, 2, 3, x, x])
```

NaN-skipping functions summarize the non-NaN entries. They do not explain why data are missing or guarantee an unbiased statistical analysis.

If every entry is NaN, `np.nansum` returns 0, whereas `np.nanmean` returns NaN and warns. An all-missing sum of zero does not establish that the underlying total was zero. These functions also do not automatically exclude infinity.

The following table provides a list of useful aggregation functions available in NumPy:

|Function Name      |   NaN-skipping Version  | Description                                   |
|-------------------|---------------------|-----------------------------------------------|
| ``np.sum``        | ``np.nansum``       | Compute sum of elements                       |
| ``np.prod``       | ``np.nanprod``      | Compute product of elements                   |
| ``np.mean``       | ``np.nanmean``      | Compute mean of elements                      |
| ``np.std``        | ``np.nanstd``       | Compute standard deviation                    |
| ``np.var``        | ``np.nanvar``       | Compute variance                              |
| ``np.min``        | ``np.nanmin``       | Find minimum value                            |
| ``np.max``        | ``np.nanmax``       | Find maximum value                            |
| ``np.argmin``     | ``np.nanargmin``    | Find index of minimum value                   |
| ``np.argmax``     | ``np.nanargmax``    | Find index of maximum value                   |
| ``np.median``     | ``np.nanmedian``    | Compute median of elements                    |
| ``np.percentile`` | ``np.nanpercentile``| Compute percentiles (q from 0 to 100)     |
| ``np.any``        | N/A                 | Evaluate whether any elements are true        |
| ``np.all``        | N/A                 | Evaluate whether all elements are true        |

`np.percentile` uses `method='linear'` by default; different methods can give different answers for small datasets. State the method when matching results across software.

For all-NaN input, `nanmin`/`nanmax` return NaN with a warning; `nanargmin`/`nanargmax` raise `ValueError`. `any`/`all` are truth tests, not missing-data checks: NaN counts as truthy.

Use `np.isnan()` to detect NaN in numeric arrays. Equality is unsuitable because NaN is unequal to every value, including itself.

```python
data = np.array([1.5, 2.3, np.nan, 4.1, np.nan, 6.0])
print(data == np.nan) # all False, including at the NaN positions
```

```python
# You must use the dedicated np.isnan() function
print(np.isnan(data))
```

```python
# Identify NaNs with np.isnan(), then negate the Boolean mask to keep other values.
clean_data = data[~np.isnan(data)]
print(clean_data)
```

## Broadcasting

Broadcasting lets compatible arrays participate in elementwise operations without explicitly repeating their data. It does **not** cycle through a shorter vector as R recycling can.

Compare shapes from **right to left**. Each pair of lengths must be equal, or one must be 1. Missing leading dimensions act like length 1.

For example, `(3, 3)` and `(3,)` are compatible; `(2, 3)` and `(2,)` are not. A shape of `(2, 1)` can broadcast with `(2, 3)`. The input arrays keep their shapes; the result may be larger.

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
print(a + b)
```

```python
c = np.array([7,8])
print(a + c)  # Intentional error: lengths 3 and 2 do not match, and neither is 1.
```

```python
print(a)
```

```python
e = np.ones([3,3])
print(e)
```

```python
print(e + a)  # the array a gets 'broadcast' across all three rows
```

```python
print(a.reshape([3,1]))  # we reshape a to be a 3x1 array
```

```python
print(e + a.reshape([3,1])) # the reshaped array is broadcast across columns
```

```python
d = np.vstack([a,b])  # we stack the arrays a and b vertically
print(d)
```

```python
a
```

```python
print(d + a)  # shape (3,) broadcasts across both rows of shape (2, 3)
```

```python
print(c)
```

```python
print(d)
```

```python
print(d + c)  # Intentional error: shapes (2, 3) and (2,) are incompatible.
```

```python
print(d + c.reshape([2,1]))  # after we reshape c to be a column, we can broadcast it
```

```python
e = np.arange(10).reshape((10, 1))
f = np.arange(11)
print(e)
print(f)
```

```python
print(e + f)  # (10, 1) + (11,) -> (10, 11), containing all pairwise sums
```

```python
print(e * f)  # (10, 1) * (11,) -> (10, 11), containing all pairwise products
```

## Combining Boolean arrays

For Boolean arrays, use `&` for elementwise AND, `|` for OR, `^` for exclusive OR, and `~` for NOT. Put parentheses around each comparison.

Python's `and`, `or`, and `not` expect the truth of a whole object and do not combine multi-element arrays elementwise. Use `np.any` or `np.all` when you want one truth value for the whole array.

```python
a = np.array([True, True, False, False])
b = np.array([True, False, True, False])
print(a)
print(b)
```

```python
print(a & b) # bitwise and
```

```python
print(a | b) # bitwise or
```

```python
print(a ^ b) # bitwise xor (exclusive or)
```

```python
print(~a)  # bitwise not
```

```python
np.any(a)
```

```python
np.all(a)
```

Python gives bitwise `&` and `|` higher precedence than comparisons. The next cell intentionally omits parentheses: it becomes the chained comparison `x > (2 & x) < 8`, which tries to use a multi-element array as one truth value.

```python
x = np.arange(10)
# Intentional error: Python gives & higher precedence than comparisons.
# This will throw a ValueError because it evaluates as x > (2 & x) < 8
x > 2 & x < 8 
```

```python
# To avoid the issue, you MUST use parentheses:
print((x > 2) & (x < 8))
```

Example:  Finding values outside a range.

```python
# Intentional error: parentheses are missing around each comparison.
# Python reads: x < (2 | x) > 7
print(x < 2 | x > 7) 
```

```python
# DO THIS
print((x < 2) | (x > 7))
```

```python
# Using it as a mask:
print(x[(x < 2) | (x > 7)])
# Output: [0 1 8 9]
```

Invert a condition.

Let's say you want to invert the output of x == 3.

```python
# DON'T DO THIS
# ~x evaluates first. 
print(~x == 3)
```

```python
~x
# For this signed integer array, bitwise inversion gives ~x = -x - 1.
# For example, ~3 is -4. Apply ~ to a Boolean mask to negate a condition;
# applying it to integer data instead flips the integer bits.
```

```python
# DO THIS
# Evaluate the equality first, THEN negate the resulting boolean array.
print(~(x == 3))
```

## Sorting values and finding their order

- `np.sort(x)` returns a sorted copy; `x.sort()` sorts in place and returns `None`.
- `np.argsort(x)` (or `x.argsort()`) returns the indices that put values in sorted order.
- Use `kind='stable'` when equal values should retain their original relative order.

```python
rng = np.random.default_rng(2)
x = np.arange(5)
rng.shuffle(x)
print(x)
```

```python
x.sort() # sorts x in place
print(x)
```

```python
y = np.array([5, 2, 1, 4])
print(y)
print(y.argsort())
```

```python
d = y.argsort()
y[d]
```

## Sorting along rows or columns

`np.sort(X, axis=0)` sorts each column independently; `axis=1` sorts each row independently (the default is the last axis).

**Independent column sorting breaks the relationships within rows.** If each row is an observation and you want to order observations by one column, use that column's `argsort` to reorder entire rows, as in the final example.

```python
rng = np.random.default_rng(1)
X = rng.integers(0, 10, size=(4, 6))
print(X)
```

```python
# Sort each column independently; np.sort returns a copy and leaves X unchanged.
np.sort(X, axis=0)
```

```python
# sort each row of X
np.sort(X, axis=1)
```

```python
X[0,:] # selecting a row
```

```python
print(X)
```

```python
X[:, 1].argsort(kind='stable')  # order rows by column index 1; preserve ties' order
```

```python
order = X[:, 1].argsort(kind='stable')
print(X[order, :])  # reorder whole rows, preserving each observation's values
```
