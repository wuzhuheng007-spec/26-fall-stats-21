Write a function `now()` that reads the current time once and prints the UTC time of day
in hours, minutes, and seconds, plus the number of complete days since the Unix epoch
(January 1, 1970, 00:00:00 UTC). It does not need to return a value.

For example, the format should look like:

```text
Current time is: 15:25:47. It has been 18370 days since the epoch.
```

Your numbers will reflect the time when you run the function, not the example above.
Use `int()` to drop the fractional seconds, and use integer division and remainders
to separate days, hours, minutes, and seconds. You do not need to determine years or months.

Build the function incrementally: first find complete days, then the time remaining
within the current day. Check that hours are between 0 and 23 and minutes and seconds
are between 0 and 59.

````python
def now():
    """Print the current UTC time and complete days since the Unix epoch."""
    seconds_since_epoch = int(time.time())
    days = seconds_since_epoch // (24 * 60 * 60)
    seconds_today = seconds_since_epoch % (24 * 60 * 60)
    hours = seconds_today // (60 * 60)
    minutes = (seconds_today % (60 * 60)) // 60
    seconds = seconds_today % 60
    print(f"Current time is: {hours:02d}:{minutes:02d}:{seconds:02d}. "
          f"It has been {days} days since the epoch.")


    
````

<https://allendowney.github.io/ThinkPython/chap06.html#id3>

The greatest common divisor (GCD) of `a` and `b` is the largest positive integer
that divides both with no remainder. Assume `a` and `b` are nonnegative integers
and at least one is positive.

When `b > 0`, let `r` be the remainder when `a` is divided by `b`. Then
`gcd(a, b) = gcd(b, r)`. When `b = 0`, the result is `a`.

Write `gcd(a, b)` to return the GCD. Use a loop or recursion; either is acceptable.

````python
def gcd(a, b):
    """Return the greatest common divisor using Euclid's algorithm."""
    while b > 0:
        a, b = b, a % b
    return a
````

## Modified Exercise 6.4 from *Think Python*, 2nd edition: Powers

Write `is_power(a, b)` to return `True` if `a = b ** k` for some nonnegative
integer `k`, and `False` otherwise. Assume `a` and `b` are integers with
`a >= 1` and `b >= 2`.

Exponent zero counts: `1` is a power of every allowed base because `b ** 0 = 1`.
Use a loop or recursion; neither approach is required. Use integer arithmetic
so that divisibility tests and intermediate values remain exact.

````python
def is_power(a, b):
    """Return whether a is a nonnegative integer power of b."""
    while a > 1:
        if a % b != 0:
            return False
        a //= b
    return True
````

## Modified Exercise 10.1: Sum possibly nested numeric data

<https://greenteapress.com/thinkpython2/thinkpython2.pdf>

Page 100.

Write `nested_sum(t)` to return the sum of all numeric values in `t`. The outer
input is a list. Its items may be integers, floats, or other lists, with lists
nested to any finite depth. Flat lists, mixed numbers and lists, and empty lists
are all valid inputs. Return `0` when there are no numeric values.

For example, `nested_sum([[1, 2], [3], [4, 5, 6]])` returns `21`.
Do not change the original list or any of its nested lists.

A loop-based or recursive solution is acceptable. For a loop-based approach, you
may use an extra list to keep track of items still to process. `isinstance(item, list)`
returns `True` when an item is a list; in this problem, a non-list item is numeric.
You do not need to handle strings, Booleans, or cyclic lists.

````python
def nested_sum(t):
    """Sum numbers in finite nested lists without modifying the input."""
    total = 0
    pending = list(reversed(t))
    while pending:
        item = pending.pop()
        if isinstance(item, list):
            pending.extend(reversed(item))
        else:
            total += item
    return total
````

````python
t = [1, 2]
print(nested_sum(t))  # Expected: 3
````

````python
t = [[1, 2], [3], [4, 5, 6]]
print(nested_sum(t))  # Expected: 21
````

## Exercise 10.2: Cumulative sums

<https://greenteapress.com/thinkpython2/thinkpython2.pdf>

Page 100.

Write `cumsum(t)` to accept **any finite iterable of numbers** and return a new list
of cumulative sums. The item at index `i` of the result is the sum of the first
`i + 1` input values, in their iteration order. For example, `cumsum([1, 2, 3])`
returns `[1, 3, 6]`. Empty input returns `[]`.

Lists, ranges, and iterators are valid inputs; all supplied values are numeric
and there are no nested lists. Do not assume the input supports indexing or `len()`.
Do not modify a list passed as input. Consuming a one-pass iterator while reading
its values is expected.

````python
def cumsum(t):
    """Return cumulative sums in iteration order.

    Do not change input lists.
    """
    result = []
    total = 0
    for value in t:
        total += value
        result.append(total)
    return result
````

````python
print(cumsum([1, 2, 3, 4]))  # Expected: [1, 3, 6, 10]
````

Two strings are anagrams if their letters can be rearranged to match exactly.
Write `is_anagram(word1, word2)` to return `True` for anagrams and `False` otherwise.
Ignore letter case and ordinary spaces using `text.replace(" ", "").lower()`.
Assume inputs contain only English letters and ordinary spaces.

Letter **counts** matter: matching lengths and the same distinct letters are not
enough. Two empty strings count as anagrams after normalization.

````python
def is_anagram(word1, word2):
    """Compare letter counts after ignoring case and ordinary spaces."""
    normalized1 = word1.replace(" ", "").lower()
    normalized2 = word2.replace(" ", "").lower()
    return sorted(normalized1) == sorted(normalized2)
````

## Exercise 10.7: Duplicate values

<https://greenteapress.com/thinkpython2/thinkpython2.pdf>

Page 101.

Write `has_duplicates(t)` to return `True` if any value appears more than once,
and `False` otherwise. It must not modify the input list. An empty list has no duplicates.

The list may mix integers and strings but will not contain nested lists. Compare
values using equality. Remember that sorting a mixed list of numbers and strings
can raise an error, as shown in class.

````python
def has_duplicates(t):
    """Return whether equal values repeat, preserving the input list."""
    seen = []
    for value in t:
        if value in seen:
            return True
        seen.append(value)
    return False
````

`.index()`, or a library search function.

Compare the target with the middle word. If they are equal, you have found it.
Otherwise search only the half where the target could still occur, excluding the
middle word that you already checked. If no candidates remain, return `False`.

Assume `word_list` is already sorted in Python's default string order and the target
and list use matching lowercase spelling. Do not change the input list. Empty lists
and one-item lists must work. The supplied file has about 114,000 words, so binary
search needs only about 17 middle-word comparisons in the worst case.

The supplied helper loads the local word list. You do not need to rewrite it.

````python
# Supplied helper: use this function without rewriting it.
def make_word_list():
    """Read the local words.txt and return its words
    in their existing sorted order.
    """
    words = []
    fin = open('words.txt')
    for line in fin:
        words.append(line.strip())
    fin.close()
    return words


t = make_word_list()
````

````python
def in_bisect(word_list, word):
    """Return whether word occurs in the sorted list using binary search."""
    low = 0
    high = len(word_list) - 1
    while low <= high:
        middle = (low + high) // 2
        if word_list[middle] == word:
            return True
        if word_list[middle] < word:
            low = middle + 1
        else:
            high = middle - 1
    return False
````

Two words form a reverse pair when one is the reverse spelling of the other.
Write `reverse_pairs(word_list)` to use your `in_bisect` function and print all
reverse pairs in the supplied list with words of **at least 6 letters**. Assume

Print each result as `(word, reversed_word)`. For two different words, include both
orientations, as in the original examples: `('animal', 'lamina')` and
`('lamina', 'animal')`. A palindrome is a valid self-pair and appears **once**, for
example `('deified', 'deified')`. You do not need to print that self-pair a second time.

To reverse a string, `word[::-1]` reads all characters with a step of `-1`.
For example, `'desserts'[::-1]` is `'stressed'`. Start by testing your function with
a small sorted trial list, then run it with the full word list `t`.

````python
# Example output includes both orientations of distinct reverse pairs:
# ('agenes', 'senega')
# ('animal', 'lamina')
# ...
# ('lamina', 'animal')
# ...
# ('senega', 'agenes')
# Palindromes appear once each, for example:
# ('deified', 'deified')
# ('redder', 'redder')
````

For a small trial list `['animal', 'deified', 'lamina', 'redder']`, the expected
pairs are the four below, in any order. You can temporarily use this list while
developing your script, but show the full-file results in your final submission.

```text
('animal', 'lamina')
('deified', 'deified')
('lamina', 'animal')
('redder', 'redder')
```

````python
def reverse_pairs(word_list):
    """Print reverse pairs of words with at least six letters."""
    for word in word_list:
        if len(word) >= 6:
            reversed_word = word[::-1]
            if in_bisect(word_list, reversed_word):
                print((word, reversed_word))
````
